```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def show_image(img, title="Image", cmap_type='gray'):
    plt.imshow(img, cmap=cmap_type)
    plt.title(title)
    plt.axis('off')
    plt.show()
```

```python
# Load the image
image_path = 'Sticker.jpg'  
image = cv2.imread(image_path)
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

show_image(gray, title="Grayscale Image")
```

```python
# Apply Gaussian Blur
blurred = cv2.GaussianBlur(gray, (5, 5), 0)
show_image(blurred, title="Blurred Image")
```

```python
# Perform edge detection using Canny
edges = cv2.Canny(blurred, threshold1=50, threshold2=150)
show_image(edges, title="Canny Edge Detection")
```

```python
# Find contours
contours, _ = cv2.findContours(edges.copy(), cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
contour_img = image.copy()
cv2.drawContours(contour_img, contours, -1, (0, 255, 0), 2)
contour_img_rgb = cv2.cvtColor(contour_img, cv2.COLOR_BGR2RGB)
plt.imshow(contour_img_rgb)
plt.title("Contours")
plt.axis('off')
plt.show()
```
