#Text Cleaning with Sample Data

This task demonstrate 3 common text cleaning steps using sample data  by the Kaggle notebook(https://www.kaggle.com/code/sudalairajkumar/getting-started-with-text-preprocessing):

1. Lowercasing
2. Removal of Punctuation
3. Removal of URLs

```python
# Sample Text Data
sample_texts = [
    "Breaking NEWS: Visit https://cnn.com for updates!!!",
    "HELLO World!!! How are YOU doing?",
    "Python is Awesome... check docs at https://python.org."
]

print("Original Texts:")
for text in sample_texts:
    print(text)
```

## 1️⃣ Lowercasing

```python
def lower_case(texts):
    return [text.lower() for text in texts]

lowercased = lower_case(sample_texts)
print("\nLowercased Texts:")
for text in lowercased:
    print(text)
```

## 2️⃣ Removal of Punctuation

```python
import string
import re

def remove_punctuation(texts):
    return [re.sub(f"[{re.escape(string.punctuation)}]", "", text) for text in texts]

no_punctuation = remove_punctuation(lowercased)
print("\nTexts without Punctuation:")
for text in no_punctuation:
    print(text)
```

## 3️⃣ Removal of URLs

```python
def remove_urls(texts):
    return [re.sub(r"http\S+|www\.\S+", "", text) for text in texts]

cleaned_texts = remove_urls(no_punctuation)
print("\nTexts without URLs:")
for text in cleaned_texts:
    print(text)
```
